```python
import nltk
nltk.download('punkt_tab')
```

## toknize document to sentences

```python
paragraph = """Natural language processing (NLP) is a field of artificial intelligence that enables computers to understand and process human language. 
It combines computational linguistics with machine learning, deep learning, and statistical modeling to analyze and generate human language.
 NLP is used in various applications such as chatbots, sentiment analysis, language translation, and information retrieval."""
sentences = nltk.sent_tokenize(paragraph)
print(sentences)
```

```python
type (sentences)
```

```python

i=0
for sentence in sentences:

    print(f"sentence:{i}", sentence)
    i += 1
```

## sentences to words

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(paragraph)
print(words)
```
