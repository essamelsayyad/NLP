## wordnet limitizer

```python
import nltk 
```

```python
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
wlem=WordNetLemmatizer()

```

```python
words = ["python", "pythoner", "pythoning", "pythoned", "pythonly","play", "playing", "played", "player","reading", "readable", "readably", "reader"]
for word in words:
    print(word + " : " + wlem.lemmatize(word, pos="v"))
```

```python
words = ["python", "pythoner", "pythoning", "pythoned", "pythonly","play", "playing", "played", "player","reading", "readable", "readably", "reader"]
for word in words:
    print(word + " : " + wlem.lemmatize(word, pos="n"))
```
