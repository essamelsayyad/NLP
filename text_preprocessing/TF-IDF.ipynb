```python
import pandas as pd
df=pd.read_csv('SMSSpamCollection.txt',sep='\t',names=['label','message'])
df.head()
```

```python
df=df.drop(columns='label')
df.head()
```

```python
df['clean_message']=df['message'].str.lower()
df.head()
```

```python

import string
string.punctuation
```

```python
def remove_punctuation(text):
    punctuation=string.punctuation
    return text.translate(str.maketrans('','',punctuation))
```

```python
df['clean_message']=df['clean_message'].apply(lambda x: remove_punctuation(x))
df.head()
```

```python
from nltk.corpus import stopwords
", ".join(stopwords.words('english'))
```

```python
STOPWORDS = set(stopwords.words('english'))
def remove_stopwords(text):
    return " ".join([word for word in text.split() if word not in STOPWORDS])
```

```python
df['clean_message']=df['message'].apply(lambda x: remove_stopwords(x))
df.head()
```

```python
from collections import Counter
word_count = Counter()
for text in df['clean_message']:
    for word in text.split():
        word_count[word] += 1
        
word_count.most_common(10)
```

```python
FREQUENT_WORDS = set(word for (word, wc) in word_count.most_common(3))
def remove_freq_words(text):
    return " ".join([word for word in text.split() if word not in FREQUENT_WORDS])
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_freq_words(x))
df.head()
```

```python

RARE_WORDS = set(word for (word, wc) in word_count.most_common()[:-10:-1])
RARE_WORDS
def remove_rare_words(text):
    return " ".join([word for word in text.split() if word not in RARE_WORDS])
```

```python
df['clean_message'] = df['clean_message'].apply(lambda x: remove_rare_words(x))
df.head()
```

```python
import re
def remove_spl_chars(text):
    text = re.sub('[^a-zA-Z0-9]', ' ', text)
    text = re.sub('\s+', ' ', text)
    return text
```

```python
df['clean_message']=df['clean_message'].apply(lambda x: remove_spl_chars(x))
df.head()
```

```python
from nltk import pos_tag
from nltk.corpus import wordnet
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
wordnet_map = {"N":wordnet.NOUN, "V": wordnet.VERB, "J": wordnet.ADJ, "R": wordnet.ADV}

def lemmatize_words(text):
    # find pos tags
    pos_text = pos_tag(text.split())
    return " ".join([lemmatizer.lemmatize(word, wordnet_map.get(pos[0], wordnet.NOUN)) for word, pos in pos_text])
```

```python
df['lemmatized_text'] = df['clean_message'].apply(lambda x: lemmatize_words(x))
df.head()
```

```python
from spellchecker import SpellChecker

spell = SpellChecker()

def correct_spellings(text):
    corrected_text = []
    misspelled_text = spell.unknown(text.split())
    
    for word in text.split():
        if word in misspelled_text:
            corrected_word = spell.correction(word)
            if corrected_word is not None:
                corrected_text.append(corrected_word)
            else:
                corrected_text.append(word)  # لو مش قادر يصحح، خلي الكلمة زي ما هي
        else:
            corrected_text.append(word)
            
    return " ".join(corrected_text)
```

```python
df['clean_message']=df['clean_message'].apply(lambda x: correct_spellings(x))
df.head()
```
