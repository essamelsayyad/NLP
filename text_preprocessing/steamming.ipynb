## steaming


```python
import nltk
```

```python
from nltk.stem import PorterStemmer
```

```python
words = ["python", "pythoner", "pythoning", "pythoned", "pythonly","play", "playing", "played", "player","reading", "readable", "readably", "reader"]
ps=PorterStemmer()
for word in words:
    print(f"{word} --> {ps.stem(word)}")
```

## Reg steammer

```python
from nltk.stem import  RegexpStemmer
rs=RegexpStemmer('ing$|s$|e$|able$', min=4)
```

```python
words = ["playing","congratulation","playable","play","playing","played","player"]
```

```python
for x in words:
    print(f"{x} --> {rs.stem(x)}")
```

```python
from nltk.stem import  RegexpStemmer
rs=RegexpStemmer('ing$|s$|e$|able$|ion$', min=5)
```

```python
for x in words:
    print(f"{x} --> {rs.stem(x)}")
```

## snowpoll stemming

```python
from nltk.stem import snowball
sws=snowball.SnowballStemmer('english')
for x in words:
    print(f"{x} --> {sws.stem(x)}")
```

```python
words=["يلعب","يلعبون","لعبت","لعبة","لعبت","لعبت","اللاعب"]
from nltk.stem import snowball
sws=snowball.SnowballStemmer('arabic')
for x in words:
    print(f"{x} --> {sws.stem(x)}")
```
